**Part 1: NUMPY ASSIGNMENT**

```python
import numpy as np

# Question 1: Create a NumPy array with marks [78, 65, 89, 56, 92] and display basic properties[cite: 1]
marks1 = np.array([78, 65, 89, 56, 92])
print("--- Question 1 ---")
print(f"Array: {marks1}")
print(f"Dimensions: {marks1.ndim}, Shape: {marks1.shape}, Size: {marks1.size}, Data Type: {marks1.dtype}")

# Question 2: Access specific marks using indexing and slicing on [72, 85, 64, 90, 76][cite: 1]
marks2 = np.array([72, 85, 64, 90, 76])
print("\n--- Question 2 ---")
print(f"Mark of the 3rd student (Index 2): {marks2[2]}")
print(f"Marks of the first 3 students (Slicing): {marks2[:3]}")
```

```python
# Question 3: Reshape 15 marks into an appropriate matrix format (5 students x 3 subjects)[cite: 1]
marks3_flat = np.array([78, 85, 90, 65, 72, 80, 88, 91, 84, 56, 62, 70, 95, 89, 92])
marks3_matrix = marks3_flat.reshape(5, 3)
print("--- Question 3 ---")
print(f"Reshaped Matrix (5x3):\n{marks3_matrix}")

# Question 4: Calculate final marks from internal and external marks arrays[cite: 1]
# Using dummy data for the two arrays to demonstrate the operation
internal_marks = np.array([35, 42, 28, 45, 30])
external_marks = np.array([45, 48, 60, 50, 55])
final_marks = internal_marks + external_marks
print("\n--- Question 4 ---")
print(f"Final Marks: {final_marks}")

# Question 5: Identify students who secured 50 marks or above using Boolean masking on [45, 78, 56, 32, 91][cite: 1]
marks5 = np.array([45, 78, 56, 32, 91])
passed = marks5[marks5 >= 50]
print("\n--- Question 5 ---")
print(f"Marks >= 50: {passed}")
```

```python
# Question 6: Calculate average marks of each student from a matrix[cite: 1]
print("--- Question 6 ---")
student_averages = np.mean(marks3_matrix, axis=1)
print(f"Average marks per student: {student_averages.round(2)}")

# Question 7: Class Performance Statistics on [67, 82, 91, 74, 58][cite: 1]
marks7 = np.array([67, 82, 91, 74, 58])
print("\n--- Question 7 ---")
print(f"Total: {np.sum(marks7)}, Average: {np.mean(marks7)}, Highest: {np.max(marks7)}, Lowest: {np.min(marks7)}, Std Dev: {np.std(marks7):.2f}")

# Question 8 & 9: Total marks in each subject and total marks by each student[cite: 1]
print("\n--- Question 8 & 9 ---")
subject_totals = np.sum(marks3_matrix, axis=0)
student_totals = np.sum(marks3_matrix, axis=1)
print(f"Total marks per subject (Axis 0): {subject_totals}")
print(f"Total marks per student (Axis 1): {student_totals}")
```

```python
# Question 10: Arrange marks [245, 278, 219, 290, 256] in order and determine ranking[cite: 1]
marks10 = np.array([245, 278, 219, 290, 256])
sorted_indices = np.argsort(marks10)[::-1]
print("--- Question 10 ---")
print(f"Ranked Marks (Descending): {marks10[sorted_indices]}")
print(f"Student Original Index Ranking: {sorted_indices}")

# Question 11: Identify unique marks in [85, 92, 85, 76, 92][cite: 1]
marks11 = np.array([85, 92, 85, 76, 92])
print("\n--- Question 11 ---")
print(f"Unique Marks: {np.unique(marks11)}")

# Question 12: Calculate average without considering missing value in [78, 85, np.nan, 92, 67][cite: 1]
marks12 = np.array([78, 85, np.nan, 92, 67])
print("\n--- Question 12 ---")
print(f"Average excluding NaN: {np.nanmean(marks12):.2f}")

# Question 13: Classify [95, 82, 74, 61, 45] into grade categories using conditional operations[cite: 1]
marks13 = np.array([95, 82, 74, 61, 45])
conditions = [marks13 >= 90, marks13 >= 75, marks13 >= 60, marks13 >= 50]
grades = ['A', 'B', 'C', 'D']
classifications = np.select(conditions, grades, default='F')
print("\n--- Question 13 ---")
print(f"Grades Classification: {classifications}")
```

```python
# Question 14: Generate marks using random number functionality and perform stats[cite: 1]
random_marks = np.random.randint(40, 100, size=5)
print("--- Question 14 ---")
print(f"Random Marks: {random_marks}")
print(f"Mean: {np.mean(random_marks)}, Max: {np.max(random_marks)}, Min: {np.min(random_marks)}")

# Question 15: Complete student performance analysis[cite: 1]
print("\n--- Question 15 ---")
overall_class_avg = np.mean(marks3_matrix)
above_avg = np.where(student_averages > overall_class_avg)[0] + 1
print(f"Class Average: {overall_class_avg:.2f}")
print(f"Students performing above class average (1-indexed): {above_avg}")
```

**Part 2: Python Assignments**

```python
# 1. Print all prime numbers between input range[cite: 2]
def print_primes(start, end):
    primes = [n for n in range(start, end + 1) if n > 1 and all(n % i != 0 for i in range(2, int(n**0.5) + 1))]
    print(f"Primes between {start} and {end}: {primes}")
print_primes(20, 50)

# 2. Factorial using recursion[cite: 2]
def factorial(n):
    return 1 if n <= 1 else n * factorial(n - 1)
print(f"\nFactorial of 5: {factorial(5)}")

# 3. Square of numbers using lambda[cite: 2]
square = lambda x: x ** 2
print(f"Square of 8: {square(8)}")

# 4. Find the second largest element in a list[cite: 2]
lst = [10, 20, 4, 45, 99, 99]
print(f"\nSecond largest in {lst}: {sorted(set(lst))[-2]}")

# 5. Count frequency of characters in a string[cite: 2]
string = "hello python"
freq = {char: string.count(char) for char in set(string)}
print(f"Character frequency: {freq}")

# 6. Calculate area of a circle using math library[cite: 2]
import math
radius = 7
print(f"\nArea of circle (r=7): {math.pi * (radius ** 2):.2f}")

# 7. Reverse a string without using built-in reverse[cite: 2]
s = "colab"
reversed_s = ""
for char in s: reversed_s = char + reversed_s
print(f"Reversed '{s}': {reversed_s}")

# 8. Remove duplicates from a list[cite: 2]
dup_list = [1, 2, 2, 3, 4, 4, 5]
print(f"\nWithout duplicates: {list(set(dup_list))}")

# 9. Merge two dictionaries[cite: 2]
dict1, dict2 = {'a': 1, 'b': 2}, {'c': 3, 'd': 4}
print(f"Merged Dictionary: {{**dict1, **dict2}}")

# 10. Fibonacci series using recursion[cite: 2]
def fibonacci(n):
    if n <= 1: return n
    return fibonacci(n-1) + fibonacci(n-2)
print(f"\nFibonacci series (7 terms): {[fibonacci(i) for i in range(7)]}")
```

**Part 3: Python Function Assignments**

```python
# 1. calculate(a, b, operation)[cite: 3]
def calculate(a, b, operation):
    ops = {'add': a+b, 'subtract': a-b, 'multiply': a*b, 'divide': a/b if b != 0 else "Div by 0"}
    return ops.get(operation, "Invalid operation")
print(f"Calculate (10, 5, 'multiply'): {calculate(10, 5, 'multiply')}")

# 2. sum_numbers(*args) accepting any number of arguments[cite: 3]
def sum_numbers(*args):
    return sum(args)
print(f"Sum of arguments (1, 2, 3, 4, 5): {sum_numbers(1, 2, 3, 4, 5)}")

# 3. employee(**args) accepting and displaying information[cite: 3]
def employee(**kwargs):
    print("\nEmployee Info:")
    for key, value in kwargs.items():
        print(f"  {key.capitalize()}: {value}")
employee(name="Alice", id=1024, department="Data Science", salary="$95,000")

# 4. remove_duplicates(lst) preserving original order[cite: 3]
def remove_duplicates_preserve_order(lst):
    seen = set()
    return [x for x in lst if not (x in seen or seen.add(x))]
print(f"\nUnique elements (Order Preserved): {remove_duplicates_preserve_order([1, 4, 2, 1, 4, 5])}")

# 5. Lambda function to sort a list of tuples based on the second element[cite: 3]
tuples_list = [(1, 5), (2, 3), (4, 1)]
sorted_tuples = sorted(tuples_list, key=lambda x: x[1])
print(f"Sorted Tuples: {sorted_tuples}")
```
